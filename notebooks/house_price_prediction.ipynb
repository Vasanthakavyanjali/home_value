# House Price Prediction — Module 1: Data Loading and Understanding

Dataset: **California Housing** from `scikit-learn`. The target `median_house_value` is measured in units of $100,000. The source dataset contains census-block-group information; it is not a list of individual houses.

This notebook loads the dataset, checks its shape, types, missing values, duplicates, and basic distributions. Data cleaning and model training are reserved for later modules.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'data_processing').exists() and (PROJECT_ROOT.parent / 'data_processing').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from data_processing.data_loader import load_housing_data, get_data_overview, save_raw_data
```

```python
housing_data = load_housing_data()
print('Dataset shape:', housing_data.shape)
housing_data.head()
```

```python
overview = get_data_overview(housing_data)
print('Rows:', overview['rows'])
print('Columns:', overview['columns'])
print('Duplicate rows:', overview['duplicate_rows'])
print('\nData types:')
housing_data.dtypes
```

```python
print('Missing values per column:')
housing_data.isna().sum()
```

```python
housing_data.describe().T.round(3)
```
